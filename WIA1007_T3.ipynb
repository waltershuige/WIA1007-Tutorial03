```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("studentrecord.csv")
df = df.dropna()
df.head()
```

```python
print("Total observations (rows):", df.shape[0])
print("Total features (columns):", df.shape[1])

print("\n==== Data Types ====")
print(df.dtypes)

print("\n==== Missing Values ====")
print(df.isnull().sum())

print("\n==== Numerical Summary ====")
print(df.describe())
```

```python
# 图1：OverallGrade 分布
plt.figure(figsize=(6,4))
sns.countplot(x="OverallGrade", data=df, order=["A","B","C","E","F"])
plt.title("Distribution of Overall Grade")
plt.show()

# 图2：Recommend Yes / No
plt.figure(figsize=(6,4))
sns.countplot(x="Recommend", data=df)
plt.title("Count of Recommend (Yes / No)")
plt.show()

# 图3：ResearchScore vs ProjectScore，按Recommend上色
plt.figure(figsize=(6,4))
sns.scatterplot(x="ResearchScore", y="ProjectScore", hue="Recommend", data=df, s=80)
plt.title("ResearchScore vs ProjectScore")
plt.show()
```

```python
print("\nAverage scores grouped by Recommend:")
print(df.groupby("Recommend")[["ResearchScore","ProjectScore"]].mean())
```
